## Why Inheritance?

Different elements of the Python language can help us in different ways with one of the main challenges in programming:

* to **break big tasks up into small, manageable pieces**.

When working with objects, **inheritance** avoids code repetition by declaring a new class as a **subclass** (or a **child**)
of some other class (the **parent** class).  The child then inherits all the methods defined in the parent class.


As an **example**, we will design a `Matrix` class, identify the overlap in functionality with the `Vector` class, and then restructure and simplify the code 
through a common parent class `Tensor`.

## Task 1: Inheritance Example.

Suppose you need to process data for **students** and **teachers**.  Both have attributes in common (like a name), while differing in behavior.  We define classes, one for student objects, and one for teacher objects, as subclasses of a common parent `Person`.

````python
class Person:
    def __init__(self, first, last):
        self.firstname = first
        self.lastname = last
    def __repr__(self):
        return f"{self.firstname} {self.lastname}"
    def chat(self):
        print(f"Hi, I'm {self}. How are you getting on?")
````

Then we can create a person, and make them chat ...

````python
john = Person("John", "Kelly")
print(john)
john.chat()
````

A `Student` is like a `Person`, except that each student **has** a student id, which can be used as email account.

````python
class Student:
    def __init__(self, first, last, number):
        self.firstname = first
        self.lastname = last
        self.student_id = number
    def __repr__(self):
        return f"{self.firstname} {self.lastname}"
    def chat(self):
        print(f"Hi, I'm {self}. How are you getting on?")
    def email(self):
        return f"{self.student_id}@uni.ie"
````

As before, we now can create a student and make them chat.  We can also get their email address.

````python
anna = Student("Anna", "Byrne", 4321)
print(anna)
print(anna.email())
anna.chat()
````

Inheritance allows us to avoid unnecessary repetition, by defining `Student` as a subclass of `Person`.  The methods `__repr__` and `chat` are then inherited, and need not be written out again; only the new attribute `student_id` and the new method `email` have to be added.  The attributes `firstname` and `lastname` are still assigned by hand -- we deal with that next.

````python
class Student(Person):
    def __init__(self, first, last, number):
        self.firstname = first
        self.lastname = last
        self.student_id = number
    def email(self):
        return f"{self.student_id}@uni.ie"
````

Objects of this `Student` class work exactly as above.

````python
anna = Student("Anna", "Byrne", 4321)
print(anna)
print(anna.email())
anna.chat()
````

The output is exactly the same as before.  The class definition has become much shorter, but seen from outside nothing has changed -- which is precisely the test that a **refactoring** has succeeded.

The `super` function allows for even more re-use of functionality from the parent class, for example in the object initialization. 

````python
class Student(Person):
    def __init__(self, first, last, number):
        super().__init__(first, last)
        self.student_id = number
    def email(self):
        return f"{self.student_id}@uni.ie"
````

Test it:

````python
anna = Student("Anna", "Byrne", 4321)
print(anna)
````

A teacher is like a person, except that a teacher has a title ...

````python
class Teacher(Person):
    def __init__(self, first, last, title):
        super().__init__(first, last)
        self.title = title
    def __repr__(self):
        return f"{self.title} {self.lastname}"
    def chat(self):
        print(f"Hi, I'm {self}. Have you done your homework?")
````

We can now create a teacher and make them chat:

````python
teacher = Teacher("Steven", "Kennedy", "Mr.")
teacher.chat()
````

The `Teacher` class defines methods `__repr__` and `chat` that the parent class `Person` already has.  This is called **method overriding**: the version found in the child class takes precedence over the one in the parent.

Note how `john.chat()` and `teacher.chat()` are the *same* instruction, carried out differently, according to the type of the object it is applied to.  This is called **polymorphism** (from the Greek for "many shapes"), and it is what makes it possible to treat a mixed list of `Person`, `Student` and `Teacher` objects uniformly.

````python
for person in [john, anna, teacher]:
    person.chat()
````

An overridden method is not lost: it can still be reached through `super`.  Inside a method, `super()` needs no arguments, as both the class and the object are known.  Outside, the two argument form `super(C, obj)` says explicitly: take the object `obj`, and start looking for the method in the parent of class `C`.

````python
super(Teacher, teacher).chat()
````

Note the mixture in this sentence: the greeting comes from `Person.chat`, but the name is still `Mr. Kennedy`, since the `f`-string in that method calls `__repr__` on the object, and the object is a `Teacher` no matter which method is running.

## Task 2: Matrix Algebra.

Recall the `Vector` class:

````python
class Vector:
    """A vector of numbers, which can be added and scaled."""

    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Vector({self.data})"
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, i):
        return self.data[i]
    
    def __iter__(self):
        return iter(self.data)
    
    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return Vector([x + y for x, y in zip(self, other)])
    
    def __rmul__(self, other):
        return Vector([other * x for x in self])
````

A **matrix** is a $2$-dimensional, rectangular collection of numerical data, like
$$
 A = \left[
  \begin{array}{ccc}
  1 & 0 & 1 \\
  2 & 1 & 1 \\
  0 & 1 & 1 \\
  1 & 1 & 2
  \end{array}
  \right].
$$
The rows of a matrix are vectors, and so are the columns. Here, the rows of $A$ are the vectors
$$
\left[
  \begin{array}{ccc}
  1 & 0 & 1 
  \end{array}
  \right],
\quad
  \left[
  \begin{array}{ccc}
  2 & 1 & 1 
  \end{array}
  \right],
\quad
  \left[
  \begin{array}{ccc}
  0 & 1 & 1 
  \end{array}
  \right],
\quad \text{and} \quad
  \left[
  \begin{array}{ccc}
   1 & 1 & 2
  \end{array}
  \right].
$$
Matrices are **like vectors**: you can add two matrices, provided they have the **same shape** (i.e., the same number of rows and the same number of columns), and you can stretch or shrink a matrix by a scalar factor.  So we should expect the methods implementing such behaviour for matrices to be very similar to the corresponding methods for vectors.  In fact, by regarding a matrix as a **list of vectors** (its rows), just like a vector is a **list of numbers**, the matrix methods can be almost identical to the vector methods.  And we can take advantage of the fact that Python already knows how to compute with `Vector`s ...

````python
class Matrix:
    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Matrix({self.data})"
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, i):
        return self.data[i]
    
    def __iter__(self):
        return iter(self.data)
    
    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return Matrix([x + y for x, y in zip(self, other)])
    
    def __rmul__(self, other):
        return Matrix([other * x for x in self])
````

Now we can create a `Matrix` object corresponding to the matrix $A$ above, as follows.

````python
ma = Matrix([
    Vector([1, 0, 1]),
    Vector([2, 1, 1]),
    Vector([0, 1, 1]),
    Vector([1, 1, 2])
])
ma
````

Compute $3A$, the $3$-multiple of $A$:

````python
3 * ma
````

Create another matrix, $B$, of the same  size as $A$ and compute the sum $A + B$:

````python
mb = Matrix([
    Vector([1, 2, 3]),
    Vector([4, 5, 6]),
    Vector([0, 0, 0]),
    Vector([7, 8, 9])
])
ma + mb
````

### Inheritance or Composition?

There are two different ways in which classes can be related, and the `Matrix` class uses both.

* A matrix **has** vectors: its `data` attribute is a list of `Vector` objects.  Building an object out of other objects like this is called **composition**.
* A matrix **is a** tensor: in the next task, `Matrix` will be defined as a subclass of a common parent class `Tensor`, and inherit its methods.  This is **inheritance**.

As a rule of thumb: use **composition** ("has a") when an object is made up of parts, and **inheritance** ("is a") when a class is a special case of another one.  Making `Matrix` a subclass of `Vector` would be a poor design: a matrix is not a special kind of vector, it is a list of them.

## Task 3: `Tensor`, a common parent class.

There is a lot of similarity between `Vector` and `Matrix`.  Moving common code into a common parent class, `Tensor`, can avoid a lot of repetition. 

A word on the name: in mathematics, a **tensor** is a more specific object than what this class provides -- ours is simply a list-like container whose entries can be added and scaled.  The loose usage follows that of libraries like `numpy`, `PyTorch` and `TensorFlow`, where a "tensor" is essentially an array of numbers with an arbitrary number of dimensions.

````python
class Tensor:
    """a common parent class for vectors and matrices"""

    # constructor
    def __init__(self, data):
        self.data = list(data)

    # list like behaviour
    def __len__(self):
        return len(self.data)

    def __iter__(self):
        return iter(self.data)

    def __getitem__(self, i):
        return self.data[i]

class Vector(Tensor):
    def __repr__(self):
        return f"Vector({self.data})"
    
    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return Vector([x + y for x, y in zip(self, other)])
    
    def __rmul__(self, other):
        return Vector([other * x for x in self])
    
class Matrix(Tensor):
    def __repr__(self):
        return f"Matrix({self.data})"
    
    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return Matrix([x + y for x, y in zip(self, other)])
    
    def __rmul__(self, other):
        return Matrix([other * x for x in self])
````

Check that all works as before:

````python
ma = Matrix([Vector([1, 0, 1]), Vector([2, 1, 1]), Vector([0, 1, 1]), Vector([1, 1, 2])])
mb = Matrix([Vector([1, 2, 3]), Vector([4, 5, 6]), Vector([0, 0, 0]), Vector([7, 8, 9])])
print(3 * ma)
print(ma + mb)
````

Each Python object has a **type**.  We can get the type of object `obj` by calling
```python

```
If `obj` is an integer then its type is `int`.
If `obj` is an instance of a class `C` then its type is `C`.
The type `t` itself is an object, with attributes and behaviour. E.g.,
```python
   t.__name__
```
is the name of type `t`, as a string.  And `t` can be called, like a function, for type conversion, or in the case of a class, to create new objects.

````python
t = type(3)
print(t)
print(t(3.14))
````

### `isinstance`.

The type of an object answers the question "what *is* this?", but with inheritance the more useful question is often "can this be used **as** a `Person`?".  This is what the built-in function `isinstance(obj, C)` answers: it is `True` if `obj` is an instance of the class `C`, **or of any of its subclasses**.

A comparison like `type(obj) == C`, in contrast, is `True` only for the class `C` itself, and `False` for every subclass.  So `isinstance` is almost always what you want when a class hierarchy is involved.

````python
print(isinstance(anna, Student), isinstance(anna, Person))
print(type(anna) == Student, type(anna) == Person)
print([p for p in [john, anna, teacher] if isinstance(p, Student)])
````

So `anna` is an instance of `Student` **and** of `Person`, but her `type` is exactly `Student`.  That is why `isinstance` is the right test as soon as subclasses are involved: a comparison of types alone would not recognize a `Student` as a `Person`.

We can use this kind of introspection in the `Vector` and `Matrix` classes to avoid even more repetition.  
More precisely, in the parent class `Tensor`, we can define a common method
```python
    def __repr__(self):
        return f"{type(self).__name__}({self.data})"
```
for the string representation, which works for both `Vector` and `Matrix`.  And we can define common methods for addition and scalar multiplication, like:
```python
    def __rmul__(self, other):
        return type(self)([other * x for x in self])
```
The simplified class hierarchy then looks as follows.

````python
class Tensor:
    """a common parent class for vectors and matrices"""

    # constructor
    def __init__(self, data):
        self.data = list(data)

    # list like behaviour
    def __len__(self):
        return len(self.data)

    def __iter__(self):
        return iter(self.data)

    def __getitem__(self, i):
        return self.data[i]

    def __repr__(self):
        return f"{type(self).__name__}({self.data})"
    
    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return type(self)([x + y for x, y in zip(self, other)])
    
    def __rmul__(self, other):
        return type(self)([other * x for x in self])

class Vector(Tensor):
    pass

class Matrix(Tensor):
    pass
````

Here, `pass` is Python's "do nothing" command.  Except for the type there is currently no difference between `Vector` and `Matrix` ...

````python
ma = Matrix([Vector([1, 0, 1]), Vector([2, 1, 1]), Vector([0, 1, 1]), Vector([1, 1, 2])])
mb = Matrix([Vector([1, 2, 3]), Vector([4, 5, 6]), Vector([0, 0, 0]), Vector([7, 8, 9])])
print(3 * ma)
print(ma + mb)
````

Once more the output is unchanged, although `Vector` and `Matrix` are now empty classes: every method they offer is inherited from `Tensor`.

## Task 5: List unpacking.

Some functions, like `print`, expect and handle an arbitrary number of arguments: the command
```python
print("a", "b", "c")
```
will print the letters `a`, `b` and `c`, separated by spaces.  But:

````python
l = ["a", "b", "c"]
print(l)
````

has a different effect.  To get the letters printed as before, you'd need to get rid of the brackets around `"a", "b", "c"` ...
That's exactly what the **list unpacking operator** `*` does, like so:

````python
print(*l)
````

`zip` is another example of such a function.  Note the difference: `zip(l)` receives **one** argument, the list `l`, and makes one short tuple out of each of its entries, whereas `zip(*l)` receives the three letters as **three** arguments, and pairs them up into a single tuple.

````python
print(list(zip(l)))
print(list(zip(*l)))
````

The **transpose**  $A^T$ of a matrix $A$ has the columns of $A$ as its rows. E.g.,
$$
 A = \left[
  \begin{array}{ccc}
  1 & 0 & 1 \\
  2 & 1 & 1 \\
  0 & 1 & 1 \\
  1 & 1 & 2
  \end{array}
  \right],\qquad
  A^T = \left[
  \begin{array}{cccc}
  1 & 2 & 0 & 1 \\
  0 & 1 & 1 & 1 \\
  1 & 1 & 1 & 2
  \end{array}
  \right].
$$

In principle, `zip` applied to the rows of $A$ yields the columns of $A$.
So we can use `zip` to compute the transpose $A^T$.
But for this to work, we need to provide the rows of $A$ as individual arguments to `zip`.
Luckily, list unpacking works for `Vector` and `Matrix` objects.

````python
print(*ma)
````

````python
for row in ma:
    print(*row)
````

Here is a function `transpose` that computes the transpose $A^T$ of a matrix $A$, using `zip` with list unpacking.

````python
def transpose(matrix):
    return Matrix([Vector(x) for x in zip(*matrix)])
````

````python
mt = transpose(ma)
for row in mt:
    print(*row)
````

Where `ma` has $4$ rows of length $3$, its transpose has $3$ rows of length $4$: rows and columns have swapped.

## Task 6: Decorators.

Rather than a **function** `transpose`, we'd like to have a **method**, `T` say, for computing the transpose of `ma` as `ma.T()`.
This we can define, inside the class `Matrix`, as follows:
```python
    def T(self):
        return type(self)([Vector(x) for x in zip(*self)])
```
But, given that the transpose of a matrix can be seen as an attribute of the matrix, it would feel more natural to simply use `ma.T` (without the pair of parentheses) for the transpose of `ma`.  This can be achieved by using a decorator like this:
```python
    @property
    def T(self):
        return type(self)([Vector(x) for x in zip(*self)])
```
There are other decorators that can be used to modify the behaviour of a function; `@property` is by far the most common one.

For now, here is the final version of the `Tensor` class hierarchy, with a `dot` method for dot products of `Vector`s and a property `T` for `Matrix` objects.

````python
class Vector(Tensor):
    def dot(self, other):  # v . v
        assert len(self) == len(other), "length mismatch"
        return sum(x * y for x, y in zip(self, other))

class Matrix(Tensor):
    @property
    def T(self):
        return type(self)([Vector(x) for x in zip(*self)])
````

With this we can compute dot products of vectors and matrix transposes.  Note that `ma[:2]`, a **slice** of the matrix `ma`, is handed on by `__getitem__` to the list `ma.data`, and thus gives a plain **list** of the first two rows -- not a `Matrix` object.  Here that is exactly what is needed, as the two rows are unpacked into `v` and `w`.

````python
ma = Matrix([Vector([1, 0, 1]), Vector([2, 1, 1]), Vector([0, 1, 1]), Vector([1, 1, 2])])
v, w = ma[:2]
print(v, w)
print(v.dot(w))
print(ma.T)
````

### Callable Objects.

Just as `__len__` makes an object behave like a list, and `__add__` makes it behave like a number, the special method `__call__` makes an object behave like a **function**: after defining `__call__(self, ...)`, an object `obj` can be applied to arguments, as in `obj(x)`.

````python
class Scale:
    """multiplication by a fixed factor, as a callable object"""

    def __init__(self, factor):
        self.factor = factor

    def __call__(self, tensor):
        return self.factor * tensor

triple = Scale(3)
print(triple(Vector([1, 0, 1])))
print(triple(ma))
````

## Task 8: Raising and Catching Errors.

In Notebook 1, errors were something that **happened** to us.  A program can also **raise** an error deliberately, with a `raise` statement, when it is asked to do something it cannot sensibly do.

````python
def average(numbers):
    if len(numbers) == 0:
        raise ValueError("average of an empty list")
    return sum(numbers) / len(numbers)

print(average([1, 2, 3]))
````

A raised error stops the program -- unless it is **caught**.  The `try` statement has a body of statements to attempt, followed by one or more `except` clauses, each naming a type of error it can deal with.  If the `try` body raises such an error, the corresponding `except` body takes over, and the program carries on.

````python
for data in [[1, 2, 3], []]:
    try:
        print("average:", average(data))
    except ValueError:
        print("cannot take the average of an empty list")
````

A parent class can use this to declare a method that every child class is expected to provide, without implementing it itself:
```python
class Instrument:
    def value(self):
        raise NotImplementedError("subclasses must implement value()")
```
Such a method acts as a **promise** about the interface of the child classes, and calling it on a child that forgot to provide its own version gives a clear error message rather than mysterious behaviour.  We will meet this pattern again in Notebook 8.
