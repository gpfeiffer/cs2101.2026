## Task 3: Arithmetic, Expressions.

Python supports standard math operations and more.

An **expression** is a valid combination of values, variables and operators that represents a new value.

````python
print(2 + 3)   # 5
print(2 * 3)   # 6
print(14 / 4)  # 3.5
print(14 // 4) # 3 (integer quotient)
print(14 % 4)  # 2 (remainder: 14 is 3 * 4 + 2)
print(2 ** 3)  # 8 (power)
````

Types matter in expressions:  the sum of two `int` values is a value of type `int`, the sum of two `float` values is a value of type `float`.

````python
2.0 + 3.0
````

### Operator Precedence.

When an expression contains several operators, the order in which they are applied is fixed by **operator precedence** -- the rule known from school as **BOMDAS**: brackets first, then orders (powers), then multiplication and division, and finally addition and subtraction.  Python follows the same convention, so a formula can usually be typed much as it is written on paper.

**Example:** Kinetic Energy.  Formula: $E = \frac12 m v^2$, here for a sliotar of mass $m = 0.115$ kg, travelling at $v = 40$ m/s (about $144$ km/h).

````python
m = 0.115   # kg
v = 40.0    # m/s
E = 1/2 * m * v**2
print("Energy:", E)
````

Here the power `v**2` is evaluated before the products, just like you would read the original formula.  Had the multiplications been carried out first, as in `(1/2 * m * v)**2`, the answer would have been `5.29` rather than `92.0`.

When in doubt, add brackets: they cost nothing, and they make the intended reading visible to whoever reads the code next.

### Built-in Functions.

Besides operators, Python provides a number of **built-in functions** that are always available, without any `import`.  You have met `print()` and `type()` already.  The ones used most often in this course are

* `round(x, n)`: the value of `x`, rounded to `n` decimal places,
* `abs(x)`: the absolute value of `x`,
* `len(s)`: the number of elements of `s` (a string here, a list in Notebook 2),
* `min(...)` and `max(...)`: the smallest and the largest of several values,
* `sum(...)`: the total of a list of values (Notebook 2).

````python
print(round(100/3, 2))              # 33.33
print(abs(-7))                      # 7
print(len("finance"))               # 7
print(min(3, 1, 2), max(3, 1, 2))   # 1 3
````

### Formatted Output.

Printing a comma separated list of values is quick, but the result of a computation with floating point numbers is often unreadable, like `1428.50925` for an amount of money.  The built-in function `round()` helps, but there is a better way.

An **f-string** is a string prefixed with the letter `f`, in which an expression enclosed in curly braces `{}` is replaced by its value.  A **format specification**, after a colon `:` inside the braces, says how to display that value.  Here `.2f` means: as a decimal number with exactly `2` digits after the point.

Writing `=` after the expression, as in `{A = }`, prints the expression itself, followed by `=` and its value.  This is handy for checking intermediate results.  A format specification can still follow, after a colon.

````python
print(f"After {t} years, the value is EUR {A:.2f}")
print(f"Unformatted, that would be {A}")
print(f"Any expression can go inside the braces: {2 * A:.2f}")
print(f"{A = :.2f}")
````

## Task 4: Programs and Functions.

A **program** is a list of instructions, usually executed step by step.

**Example:** Investment Growth.  Formula: $A = P (1 +r)^t$, as before.

````python
P = 1234                                         # I
r = 0.03                                         # I 
t = 5                                            # I 
A = P * (1 + r)**t                               # P
print("After", t, "years, the value is EUR", round(A, 2))  # O
````

This program consists of 3 parts, marked as `I` (Input), `P` (Process), `O` (Output).

Some functions, like the exponential function `exp()`, are defined in the `math` standard library.  This library needs to be imported before such a function can be used.

**Example:** Population Growth.  Formula: $P(t) = P_0 e^{rt}$.

````python
import math

P0 = 100                                     # I
r = 0.2                                      # I  
t = 10                                       # I
P = P0 * math.exp(r * t)                     # P
print(f"Population after {t} years is {P:.2f}")  # O
````

This program, too, has three parts: `I`, `P` and `O`.

## Functions and Function Calls.

A **function** is a **named** piece of reusable code with **input parameters** and a **return statement** for output.

A **function definition statement** has the general form (syntax):

```python
def function_name(parameters):
    function_body
    return result
```

So it consists of 
* a **header** line
    * starting with a **keyword** (`def`),
    * followed by the **name** given to the function,
    * the comma separated list of **parameters**, enclosed in parentheses `()`,
    * ending in a **colon** (`:`)
* and a **body** of 
    * a list of **statements**
    * including a **return** statement at the end
    * all indented by the same amount of significant whitespace.


**Example:** the square function $f(x) = x^2$.

````python
def square(x):
    x2 = x * x
    return x2
````

... or simply ...

````python
def square(x):
    return x * x
````

**Example:** Future Value $A = P (1 + r)^t$

````python
def future_value(P, r, t):
    A = P * (1 + r)**t
    return A
````

**Example:** Kinetic Energy $E = \frac12 m v^2$.

````python
def kinetic_energy(mass, velocity):
    energy = 1/2 * mass * velocity**2
    return energy
````

A **function call expression** has the form `function_name(arguments)`, i.e.,
* the function's **name** followed by
* a comma separated list of **arguments**, matching and serving as values for the parameters of the function's definition.

````python
square(5)
````

````python
future_value(1234, 0.03, 5)
````

````python
kinetic_energy(3.0, 4.0)
````

### Local Variables.

The parameters of a function, and any variables assigned in its body, are **local** to that function: they exist only while the function is running, and they have nothing to do with variables of the same name outside.  In `future_value(P, r, t)` above, the parameters `P`, `r` and `t` are not the variables `P`, `r` and `t` that were assigned earlier in this notebook.

This is what makes a function **reusable**: it communicates with the rest of the program only through its **arguments** and its **return value**.

A function whose body has no `return` statement still returns a value: the special value `None`, which stands for "nothing".

````python
P = 1000
print(future_value(1234, 0.03, 5))  # the function works with its own P
print(P)                            # the P out here is unchanged
````

## Task 6: Conditionals ...

... aka `if` statements.  Conditionals allow us to conditionally execute parts of the code.

General Form (syntax):

```python
if condition:
    body
elif other_condition:
    body
else:
    body
```


**Example:** Revenue Classifier.

````python
def revenue_class(revenue):
    if revenue < 0:
        result = "Loss"
    elif revenue > 0:
        result = "Profit"
    else:
        result = "Break Even"
    return result
````

````python
revenue_class(-5)
````

An `if` statement is a kind of **compound statement**. It consists of 
* possibly more than one **header** line, each
    * starting with a keyword (`if`, `elif` or `else`) and
    * ending with a colon (`:`)
* each followed by a **body** of one or more statements.

The first header is always the `if` clause.

Then there can be zero or more `elif` clauses, as needed.

And finally there can be one `else` clause.

The `if` clause and the `elif` clauses all have conditions attached.  Only the body under the first condition that is met will be executed, with the body of the `else` clause as default in case that none of the conditions is met.

### Conditional Expressions.

Sometimes it is a **value** that depends on a condition, rather than the flow of the program.  For this, python has the **conditional expression**

```python
value_if_true if condition else value_if_false
```

Unlike an `if` statement, this is an **expression**: it has a value, and can be used wherever a value is expected.

````python
n = 7
parity = "even" if n % 2 == 0 else "odd"
print(parity)
````

````python
def sign(x):
    return -1 if x < 0 else 1

print(sign(-5), sign(5))
````

## Recursion.

The body of a function may contain calls to other functions -- including a call to the very function that is being defined.  Such a function is called **recursive**.  For this to work, and to eventually come to an end, a recursive function needs

* at least one **base case**, which returns a value without calling itself, and
* a **recursive case**, which calls itself with arguments that are closer to a base case.

**Example:** the factorial $n! = 1 \cdot 2 \cdots n$ satisfies $0! = 1$ (the base case) and $n! = n \cdot (n-1)!$ (the recursive case).

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))   # 120 = 1 * 2 * 3 * 4 * 5
````

With a conditional expression, the same function fits on a single line.

````python
def factorial(n):
    return 1 if n == 0 else n * factorial(n - 1)

print(factorial(10))
````

## Task 7: Definite Loops ...

... aka `for` loops.

Repeat code a given number of times.

General Form (syntax):

```python
for variable in values:
    body
```


````python
for i in range(5):
    print("Step", i)
````

Here `range(5)` stands for the numbers `0,1,2,3,4`, counting from `0` and excluding `5`.  Strictly speaking it is not a list -- evaluating `range(5)` just echoes `range(0, 5)` -- but `list(range(5))` turns it into one.

With two arguments, `range(m, n)` counts from `m` instead of `0`, again excluding `n`.  A third argument gives the step size.

````python
print(list(range(5)))         # [0, 1, 2, 3, 4]
print(list(range(1, 6)))      # [1, 2, 3, 4, 5]
print(list(range(0, 10, 2)))  # [0, 2, 4, 6, 8]
````

A `for` loop statement consists of 
* a **header** line
    * starting with a keyword (`for`)
    * followed by a loop variable (`i`)
    * the keyword `in` and
    * a **list** of values (`range(5)`)
    * ending with a colon (`:`)
* followed by a **body** of one or more statements.

The body of instructions is executed once for each value in the given list, assigned to the loop variable for temporary use.


**Example:** Compound Interest, Year by Year.

````python
def compound_interest(P, r, t):
    for year in range(1, t+1):
        P = P * (1 + r)
        print("Year", year, ":", round(P, 2))
    return P
````

````python
compound_interest(1234, 0.03, 5)
````

## Task 8: Indefinite Loops ...

... aka `while` loops.

Repeat code until some condition fails.

General Form (syntax):

```python
while condition:
    body
```

A `while` statement consists of
* a **header** line that
    * starts with a keyword (`while`)
    * followed by a condition (a boolean expression)
    * and ends with a colon (`:`)
* and a **body** of one or more statements.

The body of statements is executed over and over until the condition is no longer `True`.

**Example:** How long does it take for an investment to double in value?

````python
P = 1234
r = 0.04
t = 0
A = P
while A < 2*P:
    A *= (1 + r)
    t += 1
print("doubling time:", t, "years")
````

... or simply

````python
r = 0.04
t = 0
A = 1
while A < 2:
    A *= (1 + r)
    t += 1
print("doubling time:", t, "years")
````

**Fun Fact.**  By the [Rule of 72](https://en.wikipedia.org/wiki/Rule_of_72), the doubling time at rate $r$% is roughly $72/r$.  Check: $72/4$ is $18$. $\checkmark$
(The exact formula would be $t = \ln(2)/\ln(1 +r)$.)

**Example:** Bacterial Growth up to a given Threshold.

````python
N = 100
r = 0.1
t = 0
while N < 10000:
    N *= (1 + r)
    t += 1
print("Population exceeds 10,000 after", t, "time steps")
````

## Task 9: Errors and Debugging.

Even experienced programmers make mistakes! Python will often give you feedback in the form of **error messages**. Learning to **read and interpret** error messages is an important skill.

### Common Pitfalls.

* Mixing up `==` (equals) and `=` (assignment).

* Off-by-one errors in `range(m, n)`.  This list has `n-m` elements, starting at `m` and thus does not include `n`.

* Running cells out of order.  Variables keep the values they were given in earlier cells, so a notebook that works from top to bottom can behave differently after jumping around.  When in doubt, restart the kernel and run all cells in order.

* Infinite loops, due to a condition that never fails, like:
  ```python
  V = 1
  while V < 2*V:
      V += 1
  ```

### Common Types of Errors.

1. **Syntax Errors** – mistakes in the structure of the code.

   ```python
   if x = 5:   # SyntaxError
       print("x is 5")
   ```

   Python says: `SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?`

   Correct:

   ```python
   if x == 5:  # correct
       print("x is 5")
   ```

2. **Runtime Errors** – occur when the code runs into something impossible.

   ```python
   print(10 / 0)  # ZeroDivisionError
   ```

3. **Name Errors** – using variables that don’t exist.

   ```python
   print(myVar)  # NameError: name 'myVar' is not defined
   ```

4. **Type Errors** – mixing incompatible types.

   ```python
   "abc" + 5  # TypeError
   ```




### Debugging Strategies.

* **Read the error message carefully** – Python usually tells you both the error type and the line number.

* **Use print statements** (or better, a debugger) to check variable values.

* **Test incrementally** – build code step by step instead of writing everything at once.

* **Google the error message** – most likely, someone else has had the same issue.


### Testing with `assert`.

An `assert` statement checks that a condition is `True` and raises an `AssertionError` if it is not.  When the test passes, nothing happens at all -- which is exactly what makes assertions useful: a cell full of them that runs silently confirms that every one of them holds.

````python
assert square(5) == 25
assert kinetic_energy(2.0, 4.0) == 16.0
assert revenue_class(-5) == "Loss"
print("all tests passed")
````

A failing assertion reports the test, not the reason:

```python
assert square(5) == 24  # AssertionError
```

so it pays to add a message of your own:

```python
assert square(5) == 24, "square(5) should be 25"
```

This is a good way to carry out the instruction *"test it with 2, 5 and 10"* in the exercises above.
