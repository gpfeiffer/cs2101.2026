## Task 1: Lists (Dynamic).

A **list** is a sequential collection of data.  Each item is accessible by its **position** in the list.

### Creating and Accessing Lists.

In Python, indexing starts at `0`.  A negative index counts from the end of the list.

````python
prices = [100, 102, 105, 103]
print(prices[0])   # 100
print(prices[-1])  # 103
````

### Adding and Removing Elements.

````python
prices.append(110)    # add at end
print(prices)         # [100, 102, 105, 103, 110]
````

````python
prices.remove(102)    # remove by value
print(prices)         # [100, 105, 103, 110]
````

### Iterating Through Lists.

Lists and loops go well together, naturally.

````python
for p in prices:
    print("Price:", p)
````

### Lists and Built-in Functions.

The built-in functions from Notebook 1 apply to a list as a whole: `len()` counts its entries, `sum()`, `min()` and `max()` combine them, and `sorted()` returns a **new** list with the same entries in increasing order (leaving the original list unchanged).

````python
print(prices)                     # [100, 105, 103, 110]
print(len(prices))                # 4 entries
print(sum(prices))                # 418
print(min(prices), max(prices))   # 100 110
print(sum(prices) / len(prices))  # 104.5, the average price
print(sorted(prices))             # [100, 103, 105, 110]
````

### Entries and their Positions.

A `for` loop over a list visits its **entries**.  Sometimes the **position** of an entry is needed as well: to report *where* a value was found, or to pick out the entry in the same position of another list.  The built-in function `enumerate(l)` is made for this: it turns the list `l` into a sequence of **pairs** `(i, x)`, each consisting of a position `i` and the entry `x = l[i]` in that position.  The loop `for i, x in enumerate(l):` then has both at hand.

The same effect can be had by looping over the positions alone, that is over `range(len(l))`, and using each position `i` as an index into the list.  You will see that form in other people's code, but `enumerate` is to be preferred: it keeps an entry together with its position, instead of fetching the entry back out of the list.

````python
returns = [0.01, -0.02, 0.03]
weights = [10, 20, 30]
for i, r in enumerate(returns):
    print(i, r, weights[i])
````

````python
total = 0
for i, r in enumerate(returns):
    total += r * weights[i]
print(f"weighted total: {total:.2f}")
````

## Task 2: Slices, Copies and Nested Lists.

Lists can be taken apart, they can be copied -- or accidentally not copied -- and they can contain lists themselves.  These three properties are what make lists flexible enough for the data of a real problem.

### Slicing.

A **slice** `l[start:stop]` of a list `l` is the sublist of the entries at positions `start, start+1, ...`, up to but excluding `stop`.  An omitted `start` means "from the beginning", an omitted `stop` means "to the end", so that `l[:]` is the entire list.  Negative positions count from the end.

Slicing makes a **window** of consecutive entries easy to describe: `l[i:i+3]` is the window of length `3` that starts at position `i`.

````python
data = [1, 2, 3, 4, 5]
print(data[1:3])   # [2, 3]
print(data[:3])    # [1, 2, 3]
print(data[3:])    # [4, 5]
````

````python
print(data[-2:])   # [4, 5], the last two entries
print(data[1:4])   # [2, 3, 4], a window of length 3 at position 1
````

### Copies and Aliases.

Lists are **mutable**: methods like `append()` and `remove()` modify a list in place, rather than making a new one.  An assignment does **not** copy a list, it only gives the same list another name.  Both names are then **aliases** for one and the same list, and a change made through one of them is visible through the other.

A genuine **copy** is made by `l[:]`, or by `list(l)`, or by `l.copy()`.  This is why an **immutable** type, like the tuple below, can be the safer choice: it cannot be changed behind your back.

````python
a = [1, 2, 3]
b = a          # b is just another name for the same list
b.append(4)
print(a)       # [1, 2, 3, 4]: a has changed, too!
````

````python
c = a[:]       # a genuine copy of the list
c.append(5)
print(a)       # [1, 2, 3, 4]: unchanged
print(c)       # [1, 2, 3, 4, 5]
````

### Nested Structures.

The entries of a list can be lists themselves.  A **matrix**, for example, is naturally represented as a list of **rows**, each of which is a list of numbers.  Then `m[i]` is row `i`, and the double index `m[i][j]` selects the entry in row `i` and column `j`.  Dictionaries can be nested in the same way, and so can lists and dictionaries be combined.

````python
m = [[1, 2, 3],
     [4, 5, 6]]

print(m[0])       # [1, 2, 3], the first row
print(m[1][2])    # 6, in row 1, column 2
````

````python
for row in m:
    print(row, "with row sum", sum(row))
````

## Task 3: List Comprehensions.

List comprehension provides a compact syntax for making new lists from old.

General Form (Syntax):

```python
[expression for item in list if condition]
```

A list of squares, for example:

````python
squares = [ x**2 for x in range(10) ]
print(squares)
````

Or, **even** squares only:

````python
[ x**2 for x in range(10) if x % 2 == 0 ]
````

The loop equivalent needs more space:

````python
squares = []
for x in range(10):
    if x % 2 == 0:
        squares.append(x ** 2)
print(squares)
````

## Task 4: Dictionaries.

Different tasks need different data structures.   Sometimes, a collection of key-value pairs is preferable over a list of indexed data.

In Python, the **dictionary** is the compound data structure for key-value pairs.

### Creating a dictionary, accessing values.

Dictionary entries have the general form `key : value`.  A dictionary can be specified as a comma separated list of such pairs, enclosed in curly braces.

````python
quotes = { "AAPL": 150, "GOOG": 2800, "TSLA": 700 }
quotes
````

A value in a dictionary can be accessed through its key (as opposed to its position in a list).

````python
quotes["AAPL"]
````

The items in a dictionary have no associated position.  Trying to access
```python
quotes[1]   # <- don't do this on a dictionary!!
```
would result in a `KeyError`: between the brackets, a dictionary expects a **key**, and `1` is not one of the keys here.

### Adding and updating values.

Index notation can also be used to add or update values in a dictionary.

````python
quotes["AMZN"] = 3300
quotes["AAPL"] = 155
print(quotes)
````

### Loops over dictionaries.

A `for` loop over a dictionary will loop over the keys of the dictionary.  There are methods (`items`, `keys`, `values`, ...) that make the content of a dictionary available in other ways.

````python
for key in quotes:
    print(key, "=>", quotes[key])
````

````python
for key, value in quotes.items():
    print(key, ": ", value)
````

````python
print(quotes.keys())
print(quotes.values())
````

**Example:**  Determine the value of a portfolio from two dictionaries, one containing today's prices, and one containing numbers of shares.

```python
quotes =  { "AAPL": 150, "GOOG": 2800, "TSLA": 700 }
portfolio =  { "TSLA": 5, "GOOG": 10, "AAPL": 50 }
```

````python
quotes =  { "AAPL": 150, "GOOG": 2800, "TSLA": 700 }
portfolio =  { "TSLA": 5, "GOOG": 10, "AAPL": 50 }
value = 0
for key in quotes:
    value += quotes[key] * portfolio[key]
value
````

Or, more concisely, by summing a **generator expression** -- a comprehension without the enclosing brackets:

````python
sum(quotes[key] * portfolio[key] for key in quotes)
````

## Task 5: Sets and Tuples.

In Python, a **set** is an unordered collection of data, containing each element at most once.

````python
letters = { 'b', 'a', 'n', 'a', 'n', 'a' }
letters
````

````python
print('a' in letters)
print('c' in letters)
````

### Set Operations.

The built-in function `set()` turns a list into a set, discarding duplicates.  New elements are added with the method `add()`.  Sets support the operations of set theory: **union** (`|`), **intersection** (`&`) and **difference** (`-`); the methods `union()`, `intersection()` and `difference()` do the same.  Note that a set has no order: the entries can be printed in any order.

````python
clients = ["Alice", "Bob", "Alice", "Eve"]
print(set(clients))         # each client once

evens = {2, 4, 6, 8}
odds = {1, 3, 5, 7}
print(evens | odds)         # union: all of 1, ..., 8
print(evens & {4, 8, 12})   # intersection: 4 and 8
print(evens - {4, 8})       # difference: 2 and 6

evens.add(10)
print(evens)
````

Note that the elements do not appear in the order in which they were written: `set(clients)` starts with `"Bob"`, and `evens & {4, 8, 12}` prints `{8, 4}`.  A set has **no order**, which is also the reason why its elements cannot be accessed by index.

Accessing set elements by index is not possible.  The expression

```python
letters[0]  # don't do this
```
will result in an error message.

A **tuple**, like a list, is an ordered collection of data which, unlike a list, is **immutable**.

````python
point = (-1, 3)  # x, y - coordinates
print(point)
print(point[0], ", ", point[1])
````

The components of a tuple can be accessed by index, but updating a component is not possible, the expression
```python
point[0] = 2  # don't do this
```
will result in an error message.

### Tuple Unpacking.

A tuple can be taken apart in a single assignment, by listing as many variable names on the left hand side as the tuple has components.  This is called **tuple unpacking**.

````python
point = (-1, 3)
x, y = point       # unpacking
print(x, y)

x, y = y, x        # swapping two values in one step
print(x, y)
````

Unpacking also happens in a `for` loop over a list of tuples, and that is exactly what the loop over `quotes.items()` in Task 4 does: each item is a `(key, value)` pair.

````python
points = [(0, 0), (1, 2), (3, 4)]
for x, y in points:
    print(f"{x = }, {y = }")
````

## Task 6: Comprehension, again.

The analogue of list comprehension for other compound data structures works just as you would expect.  **Dictionary comprehension**, for example:

````python
fruits = { 'apple', 'banana', 'cherry' }
{ word : len(word) for word in fruits if word[0] < 'c' }
````

Note that a **string** is a sequence, too: it can be indexed and sliced like a list, and a `for` loop (or a comprehension) over a string visits its **characters**.  Unlike a list, a string is immutable.

````python
word = "finance"
print(word[0], word[-1])   # f e
print(word[1:4])           # ina
print(len(word))           # 7
print([c for c in word])   # ['f', 'i', 'n', 'a', 'n', 'c', 'e']
````

**Set comprehension**, for example:

````python
{ c for word in fruits for c in word }
````

Note the order of the `for` loops in the above example.  The expression
```python
{ c for c in word for word in fruits }  ## don't do this
```
will result in an error message.  Why?

## Summary.

This notebook explores Python’s **core data structures**:

* **Lists**: ordered, mutable collections — useful for **sequences of data**.
* List **Comprehensions**: a concise way to create and **transform lists**.
* **Dictionaries**: key–value pairs — ideal for lookups, mappings, and **structured records**.
* **Sets**: unordered collections of unique elements — useful for **membership tests** and deduplication.
* **Slices**, **copies** and **nested** structures: sublists and windows, genuine copies as opposed to aliases, and structures built from structures.
* **Tuples**: immutable sequences — often used for **fixed records** or as dictionary keys.

Together, these structures form the backbone of data handling in Python.
They allow us to organize, access, and manipulate information efficiently, whether in scientific simulations, financial models, or real-world datasets.
