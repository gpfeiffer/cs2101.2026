## Why Objects?

Large programs need bespoke data structures.  OOP allows us to bundle data (attributes) and behaviour (methods) into developer defined data types.

As **examples**, we will develop, in a step by step fashion, a `Money` class for monetary values in a given currency, and a `Vector` class for homogeneous numerical data.

### Defining a class.

Class definitions in Python are easy.  The **general form** (syntax) of a class definition statement is
```python
class class_name:
    body
```
Like any compound statement, it 
consists of a header and a `body` of statements, all indented by the same amount of spaces.
The header, as usual, starts with a keyword (`class`) and ends with a colon(`:`), between those stands the chosen name for the class to be defined.
The body is a list of statements, usually method definitions, which mostly look like function definition statements.

A string literal as the first statement in the body of a class (or of a function) is its **docstring**, a short description of its purpose:
```python
class Money:
    """Monetary values in a given currency."""
    ...
```
It is what `help(Money)` reports, and it is available as `Money.__doc__`.  Let's see and discuss the details in the examples.

## Task 1: A `Money` class and how to create its objects.

### Defining the class.

We want to work with `Money` objects that have two **attributes**:

* `amount` - the monetary value of the object,
* `currency` - a string that denotes the currency.

As per the above description, the class definition can be as simple as

````python
class Money:
    def __init__(self, amount, currency="EUR"):
        self.amount = amount
        self.currency = currency
````

Here, the **body** of the class definition consists of a single method definition statement.

The method, `__init__()`, is a **special method**, as indicated by the double underscore (`__`) at the start and the end of its name.
It is used to create new objects of this class.

The method has three **parameters**:

* the first, by convention, is always called `self` - referring to the object in question, here the object being created.
* `amount` - corresponds to the attribute `amount`,
* `currency` - corresponds to the attribute `currency`, with **default value** `"EUR"`.

When called, the method assigns the arguments given for `amount` and `currency` to the corresponding attributes.

Note that `__init__` has no `return` statement.  By the time it runs, Python has already made the new object and passed it in as `self`; the job of `__init__` is only to fill in its attributes.  The object itself is handed back by the call `Money(104, "GBP")`.

### Creating `Money` objects.

`Money` objects are created by **calling** the class, just as you would call a function:
```python
m = Money(amount, currency)
```
Then the `__init__()` method defined as part of the `Money` class will be used to do the actual work.

````python
a = Money(104, "GBP")
b = Money(50.25)
print(f"{a.currency} {a.amount}")
print(f"{b.currency} {b.amount}")
````

## Task 2: Printing `Money` objects.

If we try to print a `Money` object, we get almost illegible noise back:  

````python
print(a)
b
````

This is python's **default** representation of an object: the name of its class, followed by the address of the object in memory (which will be different in every session).  It says nothing about the amount or the currency.

This behavior can be changed by implementing `__repr__`, another **special method** that takes care of representing an object nicely as a string.
We might, for example, have a `Money` object print itself in exactly the same way as it was created.

````python
class Money:
    def __init__(self, amount, currency="EUR"):
        self.amount = amount
        self.currency = currency

    def __repr__(self):
        return f"Money({self.amount}, {self.currency!r})"
````

With this augmented class definition we get:

````python
a = Money(104, "GBP")
b = Money(50.25)
print(a)
b
````

Two special methods take care of turning an object into text: `__repr__` aims at a **precise** representation (ideally one that could be used to recreate the object), while `__str__` aims at a **readable** one.  Simply typing `b` in a cell always shows `repr(b)`.  A `print(b)` shows `str(b)`, which falls back to `repr(b)` if the class defines no `__str__` -- as is the case here, where both lines of output look the same.

Note also the `!r` in `f"Money({self.amount}, {self.currency!r})"`: it inserts the **repr** of the value rather than the value itself, so that the currency appears in quotes, exactly as it would be typed.

If, in addition, you want to be able to print these objects in **currency code** notation, we could define our **own method** `currency_code` which does this.

````python
class Money:
    def __init__(self, amount, currency="EUR"):
        self.amount = amount
        self.currency = currency

    def __repr__(self):
        return f"Money({self.amount}, {self.currency!r})"

    def currency_code(self):
        return f"{self.currency} {self.amount}"
````

````python
a = Money(104, "GBP")
b = Money(50.25)
print(a.currency_code())
b.currency_code()
````

Note the quotation marks in the second line.  `print` writes out the text of a string, whereas a bare expression at the end of a cell is shown by its `repr`, and the `repr` of a string includes its quotes.

## Task 3: Adding `Money` objects.

The sum of two `Money` objects is a new `Money` object whose value is the sum of the values of the old `Money` objects, provided they all have the same currency.
We can implement such behavior for our `Money` class by defining the **special method** `__add__`.  
It should have two `Money` objects as its parameters, `self` and `other`, say, and return a new `Money` object according to the rules just outlined.

````python
class Money:
    def __init__(self, amount, currency="EUR"):
        self.amount = amount
        self.currency = currency

    def __repr__(self):
        return f"Money({self.amount}, {self.currency!r})"

    def currency_code(self):
        return f"{self.currency} {self.amount}"

    def __add__(self, other):
        assert self.currency == other.currency, "currency mismatch"
        return Money(self.amount + other.amount, self.currency)
````

We can now form sums of `Money` objects, but adding pounds and euros would fail with an error message.

````python
a = Money(104, "GBP")
b = Money(50.25)
c = Money(1.99)
print(b + c)
# a + b
````

### Comparing `Money` objects.

By default, `==` compares two objects by **identity**, not by value: `Money(0.3) == Money(0.3)` is `False`, because these are two *different* objects, even though they stand for the same amount of money.  Comparison by **value** is what the special method `__eq__` provides.

(Defining `__eq__` has a side effect: Python sets `__hash__` to `None`, and `Money` objects can then no longer be used as dictionary keys, or placed in a set.  A class that needs both should define `__hash__` as well.)

````python
class Money:
    """Monetary values in a given currency."""

    def __init__(self, amount, currency="EUR"):
        self.amount = amount
        self.currency = currency

    def __repr__(self):
        return f"Money({self.amount}, {self.currency!r})"

    def currency_code(self):
        return f"{self.currency} {self.amount}"

    def __add__(self, other):
        assert self.currency == other.currency, "currency mismatch"
        return Money(self.amount + other.amount, self.currency)

    def __eq__(self, other):
        return self.amount == other.amount and self.currency == other.currency
````

````python
print(Money(0.3) == Money(0.3))          # True: same amount, same currency
print(Money(0.3) == Money(0.3, "GBP"))   # False: different currency
````

Note that this `__eq__` compares a `Money` object with another `Money` object.  Something like `Money(0.3) == 5` does not return `False`; it fails with an `AttributeError`, because an `int` has no `amount` attribute.  Widening the method to cope with an operand of any type is possible -- the convention is to return the special value `NotImplemented`, which tells Python to try the comparison the other way round -- but it is more subtle than it first appears, and we leave it at that.

Now that `Money` objects are compared by value, the following comparison should be `True`.  It isn't, because floating point numbers don't always behave as you would expect:

````python
Money(0.1) + Money(0.2) == Money(0.3)
````

````python
0.1 + 0.2
````

In contrast to integer values, floating point numbers are only **approximations** to real numbers. 
As a consequence, in computations with floating point numbers we need to **watch out for rounding errors**. 

But then, are monetary values actually floating point values?  No they are not.

As an alternative design, we could store monetary values as integers: e.g., treat 5 Euro 10 cent as 510 cents, precisely because Python integers are exact -- no amount of adding them up will ever introduce a rounding error.  The conversion back to euro and cent then happens only when an amount is displayed.

Another option is `Decimal` from the standard library's `decimal` module, which represents decimal fractions **exactly**: `Decimal('0.1') + Decimal('0.2')` really is `Decimal('0.3')`.  Note the quotes -- a `Decimal` built from a string is exact, whereas `Decimal(0.1)` is handed a float that is already wrong, and faithfully keeps the error.

## Task 4: A `Vector` class for Linear Algebra.

Recall that vectors are lists of numbers that can be **added** and **scaled**.

For example, in $3$-space, if $v = (0, 1, 0)$ and $w = (2,3,4)$ then $v + w = (2,4,4)$ and $2 w = (4,6,8)$.

Plain Python lists or tuples behave differently with respect to addition and multiplication:

````python
v = (0, 1, 0)
w = (2, 3, 4)
print(v + w)
print(2*w)
````

A simple way to adjust this unwanted behaviour is to define a class of `Vector` objects, each of which refers to a list of numbers as their `data` (say) attribute, and then to implement appropriate dunder methods to achieve the desired behaviour.

Note that `__init__` below stores `list(data)` rather than `data` itself.  This makes a **copy**, so that later changes to the original list leave the `Vector` untouched -- and it turns a tuple or a `range` into a list along the way.

````python
class Vector:
    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Vector({self.data})"
````

With this class definition in place, we can already make and print `Vector` objects.

````python
v = Vector([0,1,0])
print(v)
````

## Task 5: Adding Vectors.

Now for the addition.  We can use a `for` loop to compute the vector sum of two lists, `l` and `r`, of numbers as follows.

````python
def sum_lists(l, r):
    assert len(l) == len(r), "length mismatch"
    result = []
    for i in range(len(l)):
        x = l[i]
        y = r[i]
        z = x + y
        result.append(z)
    return result
````

````python
v = [0, 1, 0]
w = [2,3,4]
sum_lists(v, w)
````

There are a couple of possible improvements to this code:

* Perhaps there is no need for the intermediate variables `x`, `y`, and `z`, as we can simply write `result.append(l[i] + r[i])`.

````python
def sum_lists(l, r):
    assert len(l) == len(r), "length mismatch"
    result = []
    for i in range(len(l)):
        result.append(l[i] + r[i])
    return result
````

````python
sum_lists(v, w)
````

* Moreover, this loop, which essentially makes a new list from two given ones, can be rephrased using **list comprehension**, thereby abolishing the need for `result` as an intermediate variable.

````python
def sum_lists(l, r):
    assert len(l) == len(r), "length mismatch"
    return [l[i] + r[i] for i in range(len(l))]
````

````python
sum_lists(v, w)
````

*  Finally, instead of, somewhat indirectly, looping over the indices in `range(len(l))` -- and `enumerate` would not help here, as the second list still has to be indexed -- we can loop directly over the given lists **in parallel**, using Python's `zip` function.

````python
def sum_lists(l, r):
    assert len(l) == len(r), "length mismatch"
    return [x + y for x, y in zip(l, r)]
````

````python
sum_lists(v, w)
````

* If we want to implement a method for addition of `Vector` objects, we can use a copy of the `sum_lists` function as definition of the **special method** `__add__` in the
`Vector` class, just taking care to rename the two parameters `l, r` to `self, other`, then find the list data in `self.data` and `other.data`, and finally to return a `Vector` object.

````python
class Vector:
    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Vector({self.data})"

    def __add__(self, other):
        assert len(self.data) == len(other.data), "length mismatch"
        return Vector([x + y for x, y in zip(self.data, other.data)])
````

Then we can add `Vector` objects as follows, by writing genuine sums.

````python
v = Vector([0, 1, 0])
w = Vector([2,3,4])
v + w
````

## Task 6: `Vector` objects as lists.

By implementing certain dunder methods, we can make our `Vector` object **behave like a list**, just as implementing `__add__` makes it **behave like a number**.
Specifically, assuming `v` is an instance of the `Vector` class:

* implementing `__len__` will make `len(v)` work,
* implementing `__getitem__` will make `v[i]` work,
* implementing `__iter__` will allow loops like `for x in v:`

It is fairly obvious what these methods should do, given that `v.data` **is** a list.

There are other benefits of list like behavior, which will eventually avoid the repeated references to the `data` attribute in `Vector` addition ...

````python
class Vector:
    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Vector({self.data})"

    def __len__(self):
        return len(self.data)

    def __getitem__(self, i):
        return self.data[i]

    def __iter__(self):
        return iter(self.data)
````

As promised, we now can treat a `Vector` object `v` as a list. 

````python
v = Vector([0, 1, 0])
print(len(v))
print(v[1])
for x in v:
    print(x)
````

The first two lines of output are `len(v)` and `v[1]`; the three lines after that come from the `for` loop, one per component.

And we can take advantage of this behaviour in the implementation of `__add__`

````python
class Vector:
    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Vector({self.data})"

    def __len__(self):
        return len(self.data)

    def __getitem__(self, i):
        return self.data[i]

    def __iter__(self):
        return iter(self.data)

    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return Vector([x + y for x, y in zip(self, other)])
````

Test it, again:

````python
v = Vector([0, 1, 0])
w = Vector([2,3,4])
v + w
````

## Task 7: Scaling `Vector` objects.

If $v$ is a vector and $s$ is a number, then the $s$ multiple of $v$ is the vector $w$ with $i$-th component $w_i = s v_i$. 
We can make `Vector` objects behave like this by implementing the **special method** `__rmul__`.
Note how the parameters `self, other` here are used to represent the product `other * self`.

````python
class Vector:
    """A vector of numbers, which can be added and scaled."""

    def __init__(self, data):
        self.data = list(data)

    def __repr__(self):
        return f"Vector({self.data})"

    def __len__(self):
        return len(self.data)

    def __getitem__(self, i):
        return self.data[i]

    def __iter__(self):
        return iter(self.data)

    def __add__(self, other):
        assert len(self) == len(other), "length mismatch"
        return Vector([x + y for x, y in zip(self, other)])

    def __rmul__(self, other):
        return Vector([other * x for x in self])
````

### Which Special Method?

Each operator corresponds to a special method, and it is always the **left** operand that is asked first.  If the left operand does not know what to do, the **right** operand is given a second chance, through the reflected method whose name carries an extra `r`.

| expression | special method | |
|---|---|---|
| `a + b` | `__add__(self, other)` | `__radd__`, if `a` cannot |
| `a - b` | `__sub__(self, other)` | `__rsub__`, if `a` cannot |
| `-a` | `__neg__(self)` | |
| `a * b` | `__mul__(self, other)` | `__rmul__`, if `a` cannot |
| `a == b` | `__eq__(self, other)` | |
| `a < b`, `a <= b` | `__lt__`, `__le__` | |
| `len(a)`, `a[i]`, `for x in a` | `__len__`, `__getitem__`, `__iter__` | |
| `print(a)`, `a` in a cell | `__str__`, `__repr__` | |

This is why `2 * v` above works with `__rmul__` alone: the left operand `2` is a number, which does not know how to multiply a `Vector`, so the `Vector` is asked next.  The expression
```python
v * 2  # don't do this, yet
```
on the other hand fails with a `TypeError` until the class also defines `__mul__` -- which is what an expression like `deposit * rate` needs.

Now we can form **linear combinations** of vectors.

````python
v = Vector([0, 1, 0])
w = Vector([2,3,4])
2 * v + 3 *  w
````
